# 01.setup the project environment for formula1 project
- create external location databricks-ext-dl1-formula1

```python
%sql
CREATE EXTERNAL LOCATION IF NOT EXISTS `databricks_ext_dl1_formula1`
URL 'abfss://formula1@databrickscourseextdl002.dfs.core.windows.net/'
WITH (STORAGE CREDENTIAL `databricks-course-sc`)
COMMENT 'formula1 project container in external location adls';
```

```python
%sql show catalogs
```

```python
%fs
ls 'abfss://formula1@databrickscourseextdl002.dfs.core.windows.net/landing'
```

```python
%sql
CREATE CATALOG formula1 
MANAGED LOCATION 'abfss://formula1@databrickscourseextdl002.dfs.core.windows.net/'
COMMENT 'this is formula1 catalog';
```

```python
%sql
CREATE SCHEMA  if not exists formula1.landing;

CREATE SCHEMA  if not exists formula1.bronze
MANAGED LOCATION 'abfss://formula1@databrickscourseextdl002.dfs.core.windows.net/bronze/';

CREATE SCHEMA  if not exists formula1.silver
MANAGED LOCATION 'abfss://formula1@databrickscourseextdl002.dfs.core.windows.net/silver/';

CREATE SCHEMA  if not exists formula1.gold
MANAGED LOCATION 'abfss://formula1@databrickscourseextdl002.dfs.core.windows.net/gold/';
```
